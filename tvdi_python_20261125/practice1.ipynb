```python
def split_bill (total, people):
    try:
        per_person = total / people
        print(f"每個人應付：{per_person:.0f}")
    except ZeroDivisionError:
        print("錯誤：分攤人數不得為 0！")
    except:
        print("其他錯誤")

split_bill(600, 3)
split_bill(600, 0)
split_bill(600, "3")
```

```python
def calc_unit_price(total_text, count_text):
    try:
        total = int(total_text)
        count = int(count_text)
        unit_price = total / count
        print(f"單價：{unit_price:.1f}元")
    except ValueError as e:
        print(f"輸入格式錯誤，請輸入整數！(系統訊息:{e})")
    except ZeroDivisionError:
        print("數量不得為 0，無法計算單價！")

calc_unit_price("500", "4")
calc_unit_price("5百", "4")
calc_unit_price("500", "0")
```
