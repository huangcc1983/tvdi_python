引數位置呼叫(positional argument)
最常見的呼叫方式，就是直接把值放進括號裡。Python 會依照呼叫時排列的先後順序，從左到右一對一對應到函式定義的參數位置

```python
def describe_pet(animal, name):
    print(f"我養了一隻{animal}, 它叫做{name}。")

# 呼叫
describe_pet("貓", "小橘")
describe_pet("小橘", "貓")
# describe_pet("貓") error, 少了引數值
# describe_pet("小橘", "貓", 3) error, 多了引數值
```

引數名稱(keyword argument)
當函式的參數比較多、不容易記住順序時，可以在呼叫時直接寫出 參數名稱=值，這種方式稱為引數名稱呼叫（或關鍵字引數）

```python
def describe_pet(animal, name):
    print(f"\n我養了一隻{animal}, 它叫做{name}。")

describe_pet(animal="貓", name="小黑")
describe_pet(name="小黑", animal="貓")
```

重要規則: 混合呼叫時，先傳位置引數，再傳名稱引數

```python
describe_pet("狗", name="旺財")

# describe_pet(animal="狗", "旺財")
```

搭配預設值（Default Arguments）
定義函式時，如果有幫參數設定預設值，呼叫時就能只改自己想調整的項目

```python
def order_drink(drink, ice="正常冰", suger="半糖"):
    print(f"點餐:{drink} {ice}/ {suger}")

order_drink("紅茶")
order_drink("紅茶", "去冰")
order_drink("紅茶", suger="無糖")
order_drink("綠茶", suger="微糖", ice="少冰")
```

特殊符號 / 與 * —— 限制引數的呼叫方式

/ (僅限位置引數 Positional-Only)
寫在 / 左邊的所有參數，呼叫時只能依位置傳入，不能寫出引數名稱

*（僅限關鍵字引數 Keyword-Only）
寫在單獨一個 * 右邊的所有參數，呼叫時必須寫出引數名稱，不能只靠位置傳入

```python
def calc_bmi(weight, height, /):
    bmi= weight / (height**2)
    print(f"BMI 為: {bmi:.1f}")

calc_bmi(65,1.75)
```

```python
def send_email(to, *, subject, urgent= False):
    print(f"寄給: {to}| 主旨: {subject}| 急件: {urgent}")

send_email("student@example.com", subject="課堂通知", urgent=True)
```

def 函式名稱( 僅限位置參數 , / , 位置或名稱皆可 , * , 僅限引數名稱 )

```python
def register(id_no, /, name, *, role="student"):
    print(f"編號：{id_no}｜姓名：{name}｜身分：{role}")
register("A01","小明",role="leader")
```

*args 與 **kwargs —— 定義函式時接收任意數量的引數（打包 Packing）
有時候設計函式時，並不知道使用者會傳入幾個引數（例如計算任意個數字的總和，或接收額外的設定欄位）。這時候可以在定義函式時，在參數前面加上 * 或 **，把多出來的引數「打包」起來。

*args（一個星號）：把多餘的「位置引數」打包成 tuple
**kwargs（兩個星號）：把多餘的「引數名稱」打包成 dict

```python
def sum_scores(*scores):
    print("收到的資料：", scores, type(scores))
    return sum(scores)
# 沒有限定數量，所以都可以
print(sum_scores(80, 90, 85))
print(sum_scores(80, 90))
print(sum_scores())
```

```python
def build_profile(name, **info):
    print("姓名：", name)
    print("其他資料：", info, type(info))

build_profile("小華",city="台北", course="Python")
```
